### Динамическое программирование

```python
# Классический пример — вычисление n-го числа Фибоначчи

# Без ДП (медленная рекурсия):
def fib(n):
    if n <= 1: return n
    return fib(n-1) + fib(n-2) # Повторяет вычисления
```

```python
fib(35)
```

```python
def fib_dp(n, memo={}):
    if n in memo: return memo[n] # Проверяем, вычисляли ли ранее
    if n <= 1: return n
    
    memo[n] = fib_dp(n-1, memo) + fib_dp(n-2, memo) # Сохраняем
    return memo[n]
```

```python
print(fib_dp(100)) # Очень быстро
```

```python
def fib_dp(n, memo={}):
    if n in memo: return memo[n] # Step 1: 2 in memo -> False; Step 2: 
    if n <= 1: return n # Step 1: 2 <= 1 -> False, 
    
    memo[n] = fib_dp(n-1, memo) + fib_dp(n-2, memo) #  Step 1: fib_dp(1, memo) + fib_dp(0, memo)
    return memo[n]

# 2
fib_dp(2)
```
