```python
def get_maximum_sum_subarray(arr, k):
    # Инициализация
    """    
    - left имеет индекс 0
    - right имеет индекс k
    - max_sum равняется sum(arr(:k)
    - current_sum, сначала равна max_sum"""

    if not arr or k == 0 or k > len(arr):
        return None

    left = 0
    right = k
    max_sum = sum(arr[left:right])
    current_sum = max_sum

    while right < len(arr):

        current_sum -= arr[left]
        current_sum += arr[right]

        max_sum = max(max_sum, current_sum)

        left += 1
        right += 1

    return max_sum
```

```python
def get_maximum_sum_subarray(arr, k):

    if not arr or k < 1 or len(arr) < k:
        return None

    left = 0
    right = k
    max_sum = sum(arr[left:right])
    curr_sum = max_sum

    while right < len(arr):

        curr_sum -= arr[left]
        curr_sum += arr[right]

        max_sum = max(max_sum, curr_sum)

        left += 1
        right += 1

    return max_sum  
```

```python
def get_maximum_sum_subarray(arr, k):

    if not arr or k <= 0 or len(arr) < k:
        return None

    left = 0
    right = k
    max_sum = sum(arr[left:right])
    current_sum = max_sum

    for i in range(k, len(arr)):
        current_sum += arr[right]
        current_sum -= arr[left]

        max_sum = max(current_sum, max_sum)

        left += 1
        right += 1
    
    return max_sum
```

```python
arr = [0, 5, -5, 1, 3, 2]
k = 3
# -> 6
assert get_maximum_sum_subarray(arr, k) == 6

arr = [0, 5, -5, 1, 3, 10]
k = 3
# -> 14
assert get_maximum_sum_subarray(arr, k) == 14

arr = []
k = 3
# пустой массив -> None
assert get_maximum_sum_subarray(arr, k) == None

arr = [1, 2, 3]
k = 4
# окно больше длины мыссива -> None
assert get_maximum_sum_subarray(arr, k) == None

arr = [1, 2, 3]
k = 0
assert get_maximum_sum_subarray(arr, k) == None

arr = [1, 2, 3]
k = 1
assert get_maximum_sum_subarray(arr, k) == 3

arr = [1, 2, 3]
k = 3
assert get_maximum_sum_subarray(arr, k) == 6

arr = [-1, -2, -3, -4]
k = 3
assert get_maximum_sum_subarray(arr, k) == -6
```
