```python
# Решить задачу со снятым ограничением на отрицательные числа ! (использование префиксных сумм и хеш-таблицы (словаря))
def has_target_sum(arr, target_sum):
    # Инициализация
    """  
    Если существует непрерывный подмассив с суммой Х, то функция вернёт True, иначе False
    Значения в массиве могут быть отрицательными

    Идея:
    Тут прямая аналогия с задачей, где
    мы искали сумму двух значений в НЕотсортированном массиве с отрицательными
    числами.
    Использовать префиксные суммы, которые будут хранится в множестве.
    И мы будем проверять для нового елемента может ли его префиксная сумма
    дополнить сумму с любым из хранимых префиксов в множестве до Х.

    Инаварианты: 
    - Не знаю

    X = -3
    arr = [
        -3, 5, -8, 10, 15
    ]
    prefix_sums = [
        -3, 2, -6, 4, 19
    ]
    -> Ок

    X = 2
    arr = [
        -3, 5, -8, 10, 15
    ]
    prefix_sums = [
        0, -3, 2, -6, 4, 19
    ]
    -> Fail

    prefix[j] - prefix[i] = target_sum ->
    prefix[i] = prefix[j] - target_sum
    """
    s = set([0]) # для инициализации самого первого элеманта
    current_sum = 0

    for i in range(len(arr)): # -3, 5

        current_sum += arr[i] # -3, 2

        if (current_sum - target_sum) in s: # 2 - 2
            return True
        else:
            s.add(current_sum)
    return False
```

```python
# Нужно вернуть (start, end) индексы подмассива с суммой X
def has_target_sum(arr, target_sum):
    # Инициализация
    """  

    prefix[j] - prefix[i] = target_sum ->
    prefix[i] = prefix[j] - target_sum
    """
    prefix_indices = {0: -1}   # префиксная сумма 0 до начала массива
    current_sum = 0
    for i, num in enumerate(arr):
        current_sum += num

        if (current_sum - target_sum) in prefix_indices:
            start = prefix_indices[current_sum - target_sum] + 1
            end = i
            return (start, end)
        
        if current_sum not in prefix_indices:   # сохраняем самый левый индекс
            prefix_indices[current_sum] = i
            
    return None
```
