```python
def has_target_sum(arr, target_sum):
    # Инициализация
    """  
    Если существует пара чисес с суммой Х, то функция вернёт True, иначе False
    """

    left = 0
    right = len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target_sum:
            return True
        elif current_sum < target_sum:
            left += 1
        else:
            right -= 1

    return False
```

```python
def has_target_sum(arr, target_sum) -> bool:

    if len(arr) <= 1:
        return False

    left = 0
    right = len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum < target_sum:
            left += 1
        elif current_sum > target_sum:
            right -= 1
        else:
            return True            

    return False
```

```python
arr = [0, 3, 7, 8, 10, 16, 21]
X = 18
assert has_target_sum(arr, X) == True

arr = [0, 3, 7, 8, 10, 16, 21]
X = 3
assert has_target_sum(arr, X) == True

arr = [0, 3, 7, 8, 10, 16, 21]
X = 37
assert has_target_sum(arr, X) == True

arr = [0, 3, 7, 8, 10, 16, 21]
X = 14
assert has_target_sum(arr, X) == False

arr = []
X = 5
assert has_target_sum(arr, X) == False

arr = [-5, 0, 3, 7, 8, 10, 16, 21]
X = 14
assert has_target_sum(arr, X) == False

arr = [-5, 0, 3, 7, 8, 10, 16, 21]
X = 2
assert has_target_sum(arr, X) == True

arr = [-5, 0, 3, 5, 7, 8, 10, 16, 21]
X = 0
assert has_target_sum(arr, X) == True
```

```python
# Если массив НЕ отсортирован -> множество
def has_target_sum(arr, target_sum) -> bool:
    
    if len(arr) < 1:
        return False

    s = set()
    for i in range(len(arr)):
        if target_sum - arr[i] not in s:
            s.add(arr[i]) 
        else:
            return True
    return False
```

```python
arr = [0, 8, 7, 3, 10, 16, 21]
X = 18
assert has_target_sum(arr, X) == True

arr = [3, 7, 8, 10, 16, 0]
X = 3
assert has_target_sum(arr, X) == True

arr = [-5, 0, 3, 7, 8, 10, 16, 21]
X = -2
assert has_target_sum(arr, X) == True

arr = [-5, 0, 3, 7, 8, 10, 5, 16, 21]
X = 0
assert has_target_sum(arr, X) == True

arr = [-5, 0, 3, 7, 8, 10, 5, 16, 21]
X = 55
assert has_target_sum(arr, X) == False
```
