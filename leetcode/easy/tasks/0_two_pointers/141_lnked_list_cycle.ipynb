```python
class ListNode(object):
    def __init__(self, x):
        self.val = x
        self.next = None

    def __repr__(self):
        return f'node: {self.val}'
```

```python
# Input: head = [3,2,0,-4], pos = 1
# Инициализация нод
node0 = ListNode(3)
node1 = ListNode(2)
node2 = ListNode(0)
node3 = ListNode(-4)
```

```python
# Создание связей между нодами
node0.next = node1
node1.next = node2
node2.next = node3

# Создаём цикл
node3.next = node1
```

```python
head = node0
```

```python
# Способ с хэш-таблицей (посещённые узлы)
def has_cycle(head: ListNode):
    visited_nodes = set()
    current = head
    while current:
        if current in visited_nodes:
            return True
        visited_nodes.add(current)
        current = current.next
    return False
```

```python
# Способ с хэш-таблицей (посещённые узлы)
def has_cycle(head: ListNode):
    visited = set()
    current = head
    while current:
        if current not in visited:
            visited.add(current)
            current.next
        else:
            return True
    return False
```

```python
assert has_cycle(head) == True
```

```python
# Алгоритм Флойда (черепаха и заяц) — два указателя с разными скоростями
def has_cycle(head: ListNode):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```

```python
# Способ с двумя указателями
def get_start_cycle(head: ListNode):
    slow = fast = head
    while fast and fast.next: # цикл обнаружения наличия петли (цикла)
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            slow = head
            while slow != fast:  # цикл синхронизации
                slow = slow.next
                fast = fast.next
            return slow
    return None
```
