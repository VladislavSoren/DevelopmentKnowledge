```python
def has_number(arr, target):
    """
    - Пользуясь этим примером, подумайте, как достичь нужного результата (стратегия)
    Здесь подойдёт бинарный поиск на основе техника two pointers двигающихся на встречу друг другу.
    Объявим два указателя left,right и будем делить массив пополам в зависимости от того в какой половине искомое значение.
    Инициализация:
    - left - имеет индекс 0
    - right - имеет индекс len(arr) - 1 (индекс последнего элемента)
    - target - целевой сумма Х, будет передаваться аргументом в функцию
    - mid - середина на которую будет делиться интервал между left и right

    Условия изменения:
    - изначально ставим mid в len(arr[left:right]) // 2
    - если arr[mid] == target, то возвращаем true
    - если arr[mid] < target, то left = mid
    - если arr[mid] > target, то right = mid
    - если left == right, но так и не нашли, то False (будем возвращать из функции по умолчанию)

    Вход:
        Х = 3
        arr - [-10, -5, 0, 3, 7]
    Вывод: True
    """
    left = 0
    right = len(arr) - 1  # 8

    if target < arr[left] or target > arr[right]:
        return False

    while left <= right:

        mid = (right + left) // 2 # 2, left(4), left(8)

        if arr[mid] > target:
            right = mid - 1
        elif arr[mid] < target:
            left = mid + 1 # 2
        else:
            return True
        
    return False
```

```python
def find_01(s):
    # Устанавливаем левую границу l на первый символ (индекс 0), который гарантированно '0'
    # Правую границу r на последний символ (индекс len(s)-1), который гарантированно '1'
    l, r = 0, len(s) - 1

    # Пока между l и r есть хотя бы один индекс (т.е. они не соседние, расстояние > 1)
    # Потому что когда l и r станут соседями (l+1 == r), то мы нашли границу перехода
    while l + 1 < r:
        # Вычисляем средний индекс между l и r (целочисленное деление)
        mid = (l + r) // 2

        # Если символ в середине — '0', значит, переход "01" находится правее mid,
        # поэтому сдвигаем левую границу в середину (l = mid), оставляя правую на месте.
        # Теперь интервал [mid, r] содержит переход.
        if s[mid] == '0':
            l = mid
        # Иначе символ в середине — '1' (другого варианта нет, так как строка из 0 и 1),
        # значит, переход находится левее mid, сдвигаем правую границу в середину (r = mid).
        else:
            r = mid

    # После цикла l и r — соседние индексы, причём s[l] == '0', s[r] == '1'.
    # Возвращаем l — индекс последнего нуля перед первой единицей.
    return l
```

```python
s = '0' * 24 + "1" * 10000
assert find_01(s) == 23
```
