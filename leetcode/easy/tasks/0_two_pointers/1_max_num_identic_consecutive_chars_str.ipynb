```python
def max_consecutive_chars(s: str) -> int:
    if not s:
        return 0
    max_len = 1
    curr_len = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            curr_len += 1
            max_len = max(max_len, curr_len)
        else:
            curr_len = 1
    return max_len
```

```python
def max_consecutive_chars(s: str) -> int:
    
    if len(s) == 0:
        return 0
    
    max_len = 1
    current_len = 1

    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            current_len += 1
        else:
            current_len = 1
        max_len = max(current_len, max_len)
    return max_len
```

```python
input_str = "q"
assert max_consecutive_chars(input_str) == 1

input_str = "qq"
assert max_consecutive_chars(input_str) == 2

input_str = "qqqqrrqqeee"
assert max_consecutive_chars(input_str) == 4

input_str = "qqrrqqeee"
assert max_consecutive_chars(input_str) == 3

input_str = "qwweerrrtt11"
assert max_consecutive_chars(input_str) == 3
```
