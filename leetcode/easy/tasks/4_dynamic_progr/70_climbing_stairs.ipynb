```python
"""
Метод динамического программирования
Очень неформально можно сказать, что метод динамического программирования заключается в том, 
чтобы найти некоторую вспомогательную величину, которая:
- помогает получить ответ на задачу,
- считается понятным образом.
Попробуем понять, что может являться такой величиной для данной задачи.
"""
```

```python
"""
Вы поднимаетесь по лестнице с n ступеньками. 
За один шаг можно подняться либо на 1 ступеньку, либо на 2. 
Сколькими разными способами можно добраться до вершины?
"""
```

```python
def climb_stairs(n): # 4
    if n == 1: return 1
    if n == 2: return 2
    db = [0] * (n + 1) # [0, 0, 0, 0, 0] строим летницу (матрицу состояний), + 1
    db[0] = 0  # если  ступенек нет, то 0 вариантов
    db[1] = 1  # если 1, то 1
    db[2] = 2  # если 1, то 2
    for i in range(3, n + 1):
        db[i] = db[i-1] + db[i-2]
    return db[n]
```

```python
climb_stairs(4)
```
